```python
import pandas as pd
# jai renommé les fichiers pour qu'ils soient plus clairs et pour que je puisse les utiliser dans le code
df = pd.read_csv("../data/rawData/2025-2026.csv")
df.head(5)
```

```python
# Columns we want to keep
wanted_columns = [
    "Date", "HomeTeam", "AwayTeam",
    "FTHG", "FTAG", "FTR",
    "AvgH", "AvgD", "AvgA",
    "AvgCH", "AvgCD", "AvgCA",
    "Avg>2.5", "Avg<2.5",
    "AvgC>2.5", "AvgC<2.5",
    "AHh", "AHCh",
    "HS", "AS", "HST", "AST", "HC", "AC"
]

df_clean = df[wanted_columns].copy()

# Add season
df_clean["Season"] = "2025-2026"

# Export cleaned dataset
df_clean.to_csv("../data/cleanedData/matches_clean2025-2026.csv", index=False)

# Display information
print("Original shape:", df.shape)
print("Cleaned shape:", df_clean.shape)
print("Columns:", df_clean.columns.tolist())
print(df_clean.head())

```

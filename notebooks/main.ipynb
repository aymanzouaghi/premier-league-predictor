```python
import pandas as pd
# jai renommé les fichiers pour qu'ils soient plus clairs et pour que je puisse les utiliser dans le code
df = pd.read_csv("../data/rawData/2025-2026.csv")
df.head(5)
```

```python
wanted_columns = [
    "Date", "HomeTeam", "AwayTeam",
    "FTHG", "FTAG", "FTR",
    "AvgH", "AvgD", "AvgA",
    "AvgCH", "AvgCD", "AvgCA",
    "Avg>2.5", "Avg<2.5",
    "AvgC>2.5", "AvgC<2.5",
    "AHh", "AHCh",
    "HS", "AS", "HST", "AST", "HC", "AC"
]
#we keep only the columns we need for our analysis in a new variable to be used in the cleaning script
```

## we clean all the data through a script and we try to see if it works 

```python
from pathlib import Path
import sys
from pathlib import Path
sys.path.append(str(Path("../src").resolve()))
from cleaning import clean_dataFile
input_dir = Path("../data/rawData")
output_dir = Path("../data/cleanedData")
all_seasons = []
for file in input_dir.glob("*.csv"):
    df_clean = clean_dataFile(file, wanted_columns)
    season = file.stem
    output_file = output_dir / f"{season}_cleaned.csv"
    df_clean.to_csv(output_file, index=False)
    all_seasons.append(df_clean)
    print(f"Cleaned data for {season} : {df_clean.shape} rows, saved to {output_file}")
```

## merging all the datasets

```python
df = pd.concat(all_seasons, ignore_index=True)  
df.shape
df.to_csv(output_dir / "all_seasons_cleaned.csv", index=False)
```
